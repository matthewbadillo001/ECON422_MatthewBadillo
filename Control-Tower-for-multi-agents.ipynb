## Step 1. Prepare the dashboard
Run the next cell once. It creates a versioned project folder beside your notebook and starts a local server. It deliberately waits until Step 2 to display the interactive-window link. Use a local Jupyter session on the same computer as your browser.

```python
# STEP 1: Prepare the included project files and start the local dashboard.
# No separate project-file downloads are needed.
from pathlib import Path
import base64, gzip, hashlib, importlib, json, re, subprocess, sys, time
from IPython.display import display, Markdown

_BUNDLED_FILES = ''
_payload = gzip.decompress(base64.b64decode(_BUNDLED_FILES))
_assets = json.loads(_payload)
APP_DIR = Path.cwd() / ("control_tower_app_" + hashlib.sha256(_payload).hexdigest()[:12])
APP_DIR.mkdir(exist_ok=True)
for _name, _text in _assets.items():
    _destination = APP_DIR / _name
    if _destination.exists() and _destination.read_text(encoding="utf-8") != _text:
        raise RuntimeError(f"Bundled file was edited: {_destination}. Rename that app folder to preserve your edits, then rerun this cell.")
    _destination.write_text(_text, encoding="utf-8")
if str(APP_DIR) not in sys.path:
    sys.path.insert(0, str(APP_DIR))
# Discard an older imported project module so all cells use this bundle.
for _module in ("simulation_support", "ml_analyst", "model_evaluation_analyst"):
    _loaded = sys.modules.get(_module)
    if _loaded is not None and Path(getattr(_loaded, "__file__", "")).parent != APP_DIR:
        del sys.modules[_module]
importlib.invalidate_caches()

_missing = [name for name in ("numpy", "pandas", "sklearn", "joblib", "matplotlib", "shap")
            if importlib.util.find_spec(name) is None]
if _missing:
    print("Missing Python packages:", ", ".join(_missing))
    print("Run this in a temporary notebook cell, then rerun Step 1:")
    print(f'%pip install -r "{APP_DIR / "requirements.txt"}"')
    raise RuntimeError("Install the listed Python packages before continuing.")

if "control_tower_process" not in globals() or control_tower_process.poll() is not None:
    dashboard_log_path = APP_DIR / "dashboard_server.log"
    with dashboard_log_path.open("w", encoding="utf-8") as dashboard_log:
        control_tower_process = subprocess.Popen(
            [sys.executable, str(APP_DIR / "dashboard.py"), "--no-browser"],
            cwd=APP_DIR, stdout=dashboard_log, stderr=subprocess.STDOUT,
        )
    for _ in range(100):
        log_text = dashboard_log_path.read_text(encoding="utf-8")
        if control_tower_process.poll() is not None:
            raise RuntimeError(log_text or "Dashboard exited before starting.")
        match = re.search(r"http://127\.0\.0\.1:\d+", log_text)
        if match:
            dashboard_url = match.group(0)
            break
        time.sleep(0.1)
    else:
        control_tower_process.terminate()
        raise RuntimeError(f"Startup timed out. Inspect {dashboard_log_path}.")
print("Step 1 complete. Edit and run the SECOND code cell next.")
print("The interactive-window link will appear there, with your DGP included.")
print("Project files and browser outputs:", APP_DIR)
# To stop this dashboard: control_tower_process.terminate()
```

### Choosing the split
Random splitting is appropriate here because the simulated rows are independent. For one real time series, set `split="time"`, specify `time_column="date"`, and choose a suitable `gap`. The helper sorts dates, uses forward CV, and reserves a final testing block. Dates must be unique. Construct lagged predictors using information available at each forecast origin and address release lags and overlapping outcomes.

Repeated units or panel data need a custom group/time-aware evaluation design; that extension is not implemented in this starter. The scripts handle continuous-outcome regression with numeric predictors.

## 5. Launch the ML Analyst
The analyst tunes the selected methods using the same configured CV folds and MSE criterion. It refits the best configuration for each method on the full estimation sample. Progress appears as each method begins and completes.

`analysis_results_path` is the location of the saved results file passed to the second analyst. It contains fitted pipelines, CV results, estimation data, and settings. It contains no testing outcomes. Random Forest uses CV here, not OOB scoring.

```python
ml_analyst = MLAnalyst(config, progress=record_progress)
record_progress("ML Analyst", "Running", "Starting the configured method comparison")
show_status()
started = perf_counter()
try:
    analysis_results_path = ml_analyst.run(X_est, y_est, RUN_DIR / "analysis")
except Exception as exc:
    record_progress("ML Analyst", "Failed", str(exc))
    raise
finally:
    status.loc["ML Analyst", "Elapsed seconds"] = round(perf_counter() - started, 2)
    status.to_csv(RUN_DIR / "control_tower_status.csv")
    show_status()
```

## 6. Inspect the ML Analyst's results
The leaderboard contains each method's best CV score and hyperparameters. CV RMSE is the square root of mean fold MSE. Fold MSE standard deviation describes variation and is not a confidence interval.

These scores both tune and select models, so they can be optimistic after selection. Final performance comes from the untouched testing sample.

```python
leaderboard = pd.read_csv(RUN_DIR / "analysis" / "leaderboard.csv")
display(leaderboard)
print("Results passed to the next analyst:", analysis_results_path.relative_to(ROOT))
record_progress("Model Evaluation Analyst", "Ready", f"{len(leaderboard)} fitted models and CV evidence received")
show_status()
```

## 7. Launch the Model Evaluation Analyst
The analyst chooses the minimum CV MSE, saves its selection, and only then evaluates testing predictions. Exact ties use model names alphabetically. A runner-up within 2% of the minimum is flagged as a descriptive near tie, not a significance test.

The analyst then computes SHAP using an estimation-sample background. Linear models use `LinearExplainer`; tree models use interventional `TreeExplainer`. Each explanation must reconstruct the prediction within numerical tolerance. Testing accuracy and SHAP plots cannot change the selected model.

```python
TOP_K = 5  # Number of important variables printed in the report
evaluation_analyst = ModelEvaluationAnalyst(
    seed=config.seed, background_size=100, max_explain=120, top_k=TOP_K,
    progress=record_progress,
)
record_progress("Model Evaluation Analyst", "Running", "Starting model review")
show_status()
started = perf_counter()
try:
    result = evaluation_analyst.run(
        analysis_results_path, X_test, y_test, RUN_DIR / "evaluation"
    )
except Exception as exc:
    record_progress("Model Evaluation Analyst", "Failed", str(exc))
    raise
finally:
    status.loc["Model Evaluation Analyst", "Elapsed seconds"] = round(perf_counter() - started, 2)
    status.to_csv(RUN_DIR / "control_tower_status.csv")
    show_status()
```

## Interpretation and scope
For observation $i$,
$$\widehat f(X_i)=\phi_0+\sum_{j=1}^p\phi_j(X_i).$$
A positive SHAP contribution raises this fitted prediction relative to the chosen background baseline. It does not establish improved accuracy or a causal effect. Mean absolute SHAP averages contribution magnitudes on the explained testing subset; it does not decompose $R^2$.

Interventional SHAP can combine correlated feature values unrealistically. The chosen background and fitted model affect attributions. This workflow does not provide causal identification, statistical tests of model superiority, or prediction intervals. Changing the model after reviewing testing outcomes requires new evaluation data for a fresh assessment.

`analysis_results.joblib` contains fitted Python objects. Only load files produced by your own trusted workflow.

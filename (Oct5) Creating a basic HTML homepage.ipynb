# Creating a Basic HTML Homepage

The goal is to learn the basic structure of an HTML webpage.

We will:

1. create a folder for the homepage;
2. write basic HTML;
3. save it as `index.html`;
4. preview it inside Jupyter.

No Git, GitHub, CSS, or deployment is used here.

## 1. Create a homepage folder

A basic website can begin with just one file:

```text
basic_homepage/
└── index.html
```

`index.html` is commonly used as the main page of a website.

````python
from pathlib import Path

# Create a folder for the homepage.
SITE_DIR = Path.cwd() / "basic_homepage"
SITE_DIR.mkdir(exist_ok=True)

# Main webpage file.
HTML_FILE = SITE_DIR / "index.html"

print("Homepage folder:")
print(SITE_DIR.resolve())
````

## 2. Basic HTML structure

A simple HTML document looks like:

```html
<html>

<head>
    information about the webpage
</head>

<body>
    visible webpage content
</body>

</html>
```

Some useful HTML tags:

| Tag | Meaning |
|---|---|
| `<html>` | Entire HTML document |
| `<head>` | Information about the page |
| `<title>` | Browser-tab title |
| `<body>` | Visible page content |
| `<h1>` | Main heading |
| `<h2>` | Section heading |
| `<p>` | Paragraph |
| `<a>` | Hyperlink |
| `<ul>` | Unordered list |
| `<li>` | List item |

````python
html = '''
<!DOCTYPE html>

<html>

<head>
    <title>My Homepage</title>
</head>

<body>

    <h1>My Homepage</h1>

    <p>
        Welcome to my homepage.
        This page was created using basic HTML.
    </p>

    <h2>About Me</h2>

    <p>
        I am interested in economics and data analysis.
    </p>

    <h2>Research Interests</h2>

    <ul>
        <li>Economics</li>
        <li>Machine Learning</li>
        <li>Data Analysis</li>
    </ul>

    <h2>Links</h2>

    <p>
        <a href="https://github.com">
            Visit GitHub
        </a>
    </p>

</body>

</html>
'''

# Save the HTML text as index.html.
HTML_FILE.write_text(html, encoding="utf-8")

print("Created:")
print(HTML_FILE.resolve())
````

### What does the code mean?

For example,

```html
<h1>My Homepage</h1>
```

creates a main heading.

```html
<p>Welcome to my homepage.</p>
```

creates a paragraph.

```html
<a href="https://github.com">Visit GitHub</a>
```

creates a hyperlink.

- `Visit GitHub` is the visible text.
- `href` contains the destination address.

A list is written as:

```html
<ul>
    <li>Economics</li>
    <li>Machine Learning</li>
</ul>
```

## 3. Preview the homepage

The browser interprets the HTML tags and displays them as a webpage.

The next cell renders the same HTML inside Jupyter.

````python
from IPython.display import HTML, display

display(HTML(html))
````


# Generate a Professional CV in HTML

This notebook creates:

```text
professional_homepage/
└── cv.html
```

The CV is designed to work with the homepage notebook.

If both notebooks are run from the same working directory, the final website folder becomes:

```text
professional_homepage/
├── index.html
├── style.css
└── cv.html
```

The homepage's **CV** button can then open this file.

## USER SPECIFICATION BOX

Edit the information below.

The same theme names used in the homepage notebook are available here:

```text
classic_blue
modern_slate
forest
burgundy
warm_sand
midnight
```

````python
# ============================================================
#                 USER SPECIFICATION BOX
# ============================================================

NAME = "Your Name"

TITLE = "Economist | Researcher | Instructor"

EMAIL = "your.email@example.com"

WEBSITE = "https://your-username.github.io/"
WEBSITE_LABEL = "your-username.github.io"

GITHUB_URL = "https://github.com/your-username"
GITHUB_LABEL = "github.com/your-username"


# ---------------------------
# Education
# ---------------------------

EDUCATION = [
    (
        "Ph.D. in Economics",
        "University Name",
        "2026"
    ),
    (
        "M.A. in Economics",
        "University Name",
        "2021"
    ),
]


# ---------------------------
# Academic / professional positions
# ---------------------------

POSITIONS = [
    (
        "Researcher / Instructor",
        "Institution Name",
        "2026-present"
    ),
]


# ---------------------------
# Research interests
# ---------------------------

RESEARCH_INTERESTS = [
    "Econometrics",
    "Machine Learning",
    "Causal Inference",
    "Applied Economics",
]


# ---------------------------
# Publications
# ---------------------------

PUBLICATIONS = [
    (
        "Paper Title",
        "Journal or working-paper information"
    ),
    (
        "Working Paper Title",
        "Working paper"
    ),
]


# ---------------------------
# Teaching
# ---------------------------

TEACHING = [
    (
        "Macroeconomics and Machine Learning",
        "University Name, 2026"
    ),
    (
        "Econometrics",
        "University Name, 2026"
    ),
]


# ---------------------------
# Awards / service
# ---------------------------

AWARDS = [
    "Award or fellowship, 2026",
]

SERVICE = [
    "Professional or university service"
]


# ---------------------------
# Design
# ---------------------------

THEME = "classic_blue"

FONT_FAMILY = "Arial, Helvetica, sans-serif"

CV_WIDTH = "820px"


# ============================================================
#               END USER SPECIFICATION BOX
# ============================================================
````

## 1. Select the theme

````python
THEMES = {

    "classic_blue": {
        "accent": "#1f4e79",
        "text": "#1f2933",
        "muted": "#667085",
        "background": "#ffffff",
        "border": "#d8dee8",
    },

    "modern_slate": {
        "accent": "#475569",
        "text": "#1f2937",
        "muted": "#64748b",
        "background": "#f8fafc",
        "border": "#cbd5e1",
    },

    "forest": {
        "accent": "#2f6b4f",
        "text": "#1f2d27",
        "muted": "#66736c",
        "background": "#fbfdfb",
        "border": "#d7e3dc",
    },

    "burgundy": {
        "accent": "#7a284b",
        "text": "#2b2025",
        "muted": "#75636b",
        "background": "#fffafa",
        "border": "#ead5de",
    },

    "warm_sand": {
        "accent": "#9a6b2f",
        "text": "#2f2a24",
        "muted": "#776f64",
        "background": "#fffdf8",
        "border": "#e9dfcf",
    },

    "midnight": {
        "accent": "#7dd3fc",
        "text": "#e5e7eb",
        "muted": "#aab4c0",
        "background": "#111827",
        "border": "#374151",
    },
}

if THEME not in THEMES:
    raise ValueError(
        f"Unknown theme: {THEME}. "
        f"Choose one of: {list(THEMES)}"
    )

theme = THEMES[THEME]

ACCENT_COLOR = theme["accent"]
TEXT_COLOR = theme["text"]
MUTED_TEXT_COLOR = theme["muted"]
BACKGROUND_COLOR = theme["background"]
BORDER_COLOR = theme["border"]

print("Theme:", THEME)
````

## 2. Create the output folder

````python
from pathlib import Path
import html as html_lib

SITE_DIR = Path.cwd() / "professional_homepage"
SITE_DIR.mkdir(exist_ok=True)

CV_FILE = SITE_DIR / "cv.html"

print("CV file:")
print(CV_FILE.resolve())
````

## 3. Convert the specification box into HTML

````python
def esc(x):
    return html_lib.escape(str(x))


education_html = ""

for degree, institution, year in EDUCATION:
    education_html += f'''
        <div class="entry">
            <div>
                <strong>{esc(degree)}</strong><br>
                <span>{esc(institution)}</span>
            </div>
            <div class="date">{esc(year)}</div>
        </div>
    '''


positions_html = ""

for position, institution, years in POSITIONS:
    positions_html += f'''
        <div class="entry">
            <div>
                <strong>{esc(position)}</strong><br>
                <span>{esc(institution)}</span>
            </div>
            <div class="date">{esc(years)}</div>
        </div>
    '''


research_html = "".join(
    f"<li>{esc(item)}</li>"
    for item in RESEARCH_INTERESTS
)


publications_html = ""

for title, info in PUBLICATIONS:
    publications_html += f'''
        <li>
            <strong>{esc(title)}</strong><br>
            <span>{esc(info)}</span>
        </li>
    '''


teaching_html = ""

for course, info in TEACHING:
    teaching_html += f'''
        <li>
            <strong>{esc(course)}</strong><br>
            <span>{esc(info)}</span>
        </li>
    '''


awards_html = "".join(
    f"<li>{esc(item)}</li>"
    for item in AWARDS
)


service_html = "".join(
    f"<li>{esc(item)}</li>"
    for item in SERVICE
)
````

## 4. Generate `cv.html`

````python
cv_html = f'''
<!DOCTYPE html>

<html lang="en">

<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">

    <title>{esc(NAME)} | CV</title>

    <style>

        * {{
            box-sizing: border-box;
        }}

        body {{
            margin: 0;
            font-family: {FONT_FAMILY};
            line-height: 1.55;
            color: {TEXT_COLOR};
            background: {BACKGROUND_COLOR};
        }}

        .cv {{
            width: min({CV_WIDTH}, 90%);
            margin: 48px auto;
        }}

        header {{
            border-bottom: 2px solid {ACCENT_COLOR};
            padding-bottom: 18px;
            margin-bottom: 32px;
        }}

        header h1 {{
            margin: 0;
            font-size: 2.15rem;
        }}

        .title {{
            margin: 5px 0;
            color: {MUTED_TEXT_COLOR};
        }}

        .contact {{
            margin-top: 9px;
            font-size: 0.92rem;
        }}

        a {{
            color: {ACCENT_COLOR};
        }}

        section {{
            margin: 30px 0;
        }}

        section h2 {{
            margin: 0 0 14px;
            padding-bottom: 5px;

            border-bottom: 1px solid {BORDER_COLOR};

            color: {ACCENT_COLOR};

            font-size: 1.05rem;
            text-transform: uppercase;
            letter-spacing: 0.06em;
        }}

        .entry {{
            display: flex;
            justify-content: space-between;
            gap: 30px;
            margin: 12px 0;
        }}

        .entry span,
        li span {{
            color: {MUTED_TEXT_COLOR};
        }}

        .date {{
            min-width: 115px;
            text-align: right;
            color: {MUTED_TEXT_COLOR};
        }}

        ul {{
            margin: 0;
            padding-left: 20px;
        }}

        li {{
            margin: 9px 0;
        }}


        /* Good behavior when printing / saving as PDF */

        @media print {{

            body {{
                background: white;
                color: black;
            }}

            .cv {{
                width: 100%;
                margin: 0;
            }}

            a {{
                color: black;
                text-decoration: none;
            }}

            section h2 {{
                color: black;
            }}
        }}


        /* Small screens */

        @media (max-width: 650px) {{

            .entry {{
                flex-direction: column;
                gap: 2px;
            }}

            .date {{
                text-align: left;
            }}
        }}

    </style>

</head>


<body>

    <main class="cv">

        <header>

            <h1>{esc(NAME)}</h1>

            <p class="title">{esc(TITLE)}</p>

            <p class="contact">

                <a href="mailto:{esc(EMAIL)}">
                    {esc(EMAIL)}
                </a>

                &nbsp; | &nbsp;

                <a href="{esc(WEBSITE)}">
                    {esc(WEBSITE_LABEL)}
                </a>

                &nbsp; | &nbsp;

                <a href="{esc(GITHUB_URL)}">
                    {esc(GITHUB_LABEL)}
                </a>

            </p>

        </header>


        <section>

            <h2>Education</h2>

            {education_html}

        </section>


        <section>

            <h2>Positions</h2>

            {positions_html}

        </section>


        <section>

            <h2>Research Interests</h2>

            <ul>
                {research_html}
            </ul>

        </section>


        <section>

            <h2>Publications and Working Papers</h2>

            <ul>
                {publications_html}
            </ul>

        </section>


        <section>

            <h2>Teaching</h2>

            <ul>
                {teaching_html}
            </ul>

        </section>


        <section>

            <h2>Awards and Fellowships</h2>

            <ul>
                {awards_html}
            </ul>

        </section>


        <section>

            <h2>Service</h2>

            <ul>
                {service_html}
            </ul>

        </section>

    </main>

</body>

</html>
'''

CV_FILE.write_text(cv_html, encoding="utf-8")

print("Created:")
print(CV_FILE.resolve())
````

## 5. Preview the CV

````python
from IPython.display import HTML, display

display(HTML(cv_html))
````

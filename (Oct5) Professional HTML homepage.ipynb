# Professional HTML Homepage with Theme Presets

This notebook generates:

```text
professional_homepage/
├── index.html
└── style.css
```

If a `cv.html` file is also placed in the same folder, the **CV** button will open it automatically.

For beginners, the easiest workflow is:

```text
edit USER SPECIFICATION BOX
        ↓
choose a theme name
        ↓
rerun notebook
        ↓
index.html + style.css
```

## USER SPECIFICATION BOX

For most changes, edit only the next cell.

### Available themes

| Theme name | Style |
|---|---|
| `classic_blue` | traditional academic / professional |
| `modern_slate` | neutral and modern |
| `forest` | muted green |
| `burgundy` | warm academic |
| `warm_sand` | soft beige / brown |
| `midnight` | dark professional |

Example:

```python
THEME = "burgundy"
```

````python
# ============================================================
#                 USER SPECIFICATION BOX
# ============================================================

# ---------------------------
# Basic information
# ---------------------------

NAME = "Your Name"

SUBTITLE = "Economist | Researcher | Instructor"

WELCOME_HEADING = "Research in economics, data, and machine learning"

ABOUT = (
    "I am an economist interested in empirical research, "
    "econometrics, and machine learning. "
    "This homepage provides information about my research, "
    "teaching, and related academic work."
)

EMAIL = "your.email@example.com"

GITHUB_URL = "https://github.com/your-username"
GITHUB_LABEL = "github.com/your-username"

COPYRIGHT_YEAR = 2026


# ---------------------------
# Homepage content
# ---------------------------

RESEARCH = [
    (
        "Econometrics",
        "Research on statistical and econometric methods "
        "for economic applications."
    ),
    (
        "Machine Learning",
        "Applications of machine learning to prediction, "
        "measurement, and causal analysis."
    ),
    (
        "Applied Economics",
        "Empirical work using economic and policy data."
    ),
]

TEACHING = [
    (
        "Macroeconomics and Machine Learning",
        "Python, machine learning, web data collection, "
        "and economic applications."
    ),
    (
        "Econometrics",
        "Regression, causal inference, and empirical methods."
    ),
]

PUBLICATIONS = [
    (
        "Paper Title",
        "Short description of the paper."
    ),
    (
        "Working Paper Title",
        "Short description of the project."
    ),
]


# ---------------------------
# Easy design choices
# ---------------------------

THEME = "classic_blue"

# Common font choices:
# "Arial, Helvetica, sans-serif"
# "Georgia, serif"
# "'Trebuchet MS', sans-serif"
FONT_FAMILY = "Arial, Helvetica, sans-serif"

PAGE_WIDTH = "960px"

CARD_COLUMNS = 3

# "right" or "left"
MENU_SIDE = "right"


# ============================================================
#               END USER SPECIFICATION BOX
# ============================================================
````

## 1. Theme presets

The user chooses only a theme name.

The notebook translates that name into a coordinated set of colors.

````python
THEMES = {

    "classic_blue": {
        "accent": "#1f4e79",
        "text": "#1f2933",
        "muted": "#667085",
        "background": "#ffffff",
        "card": "#f7f9fc",
        "border": "#d8dee8",
    },

    "modern_slate": {
        "accent": "#475569",
        "text": "#1f2937",
        "muted": "#64748b",
        "background": "#f8fafc",
        "card": "#ffffff",
        "border": "#cbd5e1",
    },

    "forest": {
        "accent": "#2f6b4f",
        "text": "#1f2d27",
        "muted": "#66736c",
        "background": "#fbfdfb",
        "card": "#f2f7f4",
        "border": "#d7e3dc",
    },

    "burgundy": {
        "accent": "#7a284b",
        "text": "#2b2025",
        "muted": "#75636b",
        "background": "#fffafa",
        "card": "#fcf4f7",
        "border": "#ead5de",
    },

    "warm_sand": {
        "accent": "#9a6b2f",
        "text": "#2f2a24",
        "muted": "#776f64",
        "background": "#fffdf8",
        "card": "#faf6ed",
        "border": "#e9dfcf",
    },

    "midnight": {
        "accent": "#7dd3fc",
        "text": "#e5e7eb",
        "muted": "#aab4c0",
        "background": "#111827",
        "card": "#1f2937",
        "border": "#374151",
    },
}

if THEME not in THEMES:
    raise ValueError(
        f"Unknown theme: {THEME}. "
        f"Choose one of: {list(THEMES)}"
    )

theme = THEMES[THEME]

ACCENT_COLOR = theme["accent"]
TEXT_COLOR = theme["text"]
MUTED_TEXT_COLOR = theme["muted"]
BACKGROUND_COLOR = theme["background"]
CARD_BACKGROUND = theme["card"]
BORDER_COLOR = theme["border"]

print("Theme:", THEME)
print(theme)
````

## 2. Create the website folder

````python
from pathlib import Path
import html as html_lib

SITE_DIR = Path.cwd() / "professional_homepage"
SITE_DIR.mkdir(exist_ok=True)

HTML_FILE = SITE_DIR / "index.html"
CSS_FILE = SITE_DIR / "style.css"

print(SITE_DIR.resolve())
````

## 3. Convert the user information into HTML blocks

````python
def esc(x):
    return html_lib.escape(str(x))


research_html = ""

for title, description in RESEARCH:
    research_html += f'''
        <article class="card">
            <h3>{esc(title)}</h3>
            <p>{esc(description)}</p>
        </article>
    '''


teaching_html = ""

for course, description in TEACHING:
    teaching_html += f'''
        <li>
            <strong>{esc(course)}</strong><br>
            {esc(description)}
        </li>
    '''


publications_html = ""

for title, description in PUBLICATIONS:
    publications_html += f'''
        <li>
            <strong>{esc(title)}</strong><br>
            {esc(description)}
        </li>
    '''
````

## 4. Build the homepage

The **CV** button opens `cv.html` in a new tab:

```html
<a href="cv.html" target="_blank">CV</a>
```

````python
html = f'''
<!DOCTYPE html>

<html lang="en">

<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">

    <title>{esc(NAME)} | Homepage</title>

    <link rel="stylesheet" href="style.css">
</head>

<body>

    <nav class="side-nav">
        <a href="#about">About</a>
        <a href="#research">Research</a>
        <a href="#teaching">Teaching</a>
        <a href="#publications">Publications</a>
        <a href="#contact">Contact</a>
    </nav>


    <header>
        <div class="container">

            <h1>{esc(NAME)}</h1>

            <p class="subtitle">
                {esc(SUBTITLE)}
            </p>

        </div>
    </header>


    <main class="container">

        <section id="about" class="hero">

            <div class="hero-text">

                <p class="eyebrow">Welcome</p>

                <h2>{esc(WELCOME_HEADING)}</h2>

                <p>{esc(ABOUT)}</p>

                <p>
                    <a class="button" href="cv.html" target="_blank">
                        CV
                    </a>
                </p>

            </div>

        </section>


        <section id="research">

            <h2>Research</h2>

            <div class="card-grid">
                {research_html}
            </div>

        </section>


        <section id="teaching">

            <h2>Teaching</h2>

            <ul class="clean-list">
                {teaching_html}
            </ul>

        </section>


        <section id="publications">

            <h2>Selected Publications and Working Papers</h2>

            <ul class="clean-list">
                {publications_html}
            </ul>

        </section>


        <section id="contact">

            <h2>Contact</h2>

            <p>
                Email:
                <a href="mailto:{esc(EMAIL)}">
                    {esc(EMAIL)}
                </a>
            </p>

            <p>
                GitHub:
                <a href="{esc(GITHUB_URL)}" target="_blank">
                    {esc(GITHUB_LABEL)}
                </a>
            </p>

        </section>

    </main>


    <footer>
        <div class="container">
            <p>© {COPYRIGHT_YEAR} {esc(NAME)}</p>
        </div>
    </footer>

</body>

</html>
'''

HTML_FILE.write_text(html, encoding="utf-8")

print("Created:")
print(HTML_FILE.resolve())
````

## 5. Generate CSS from the selected theme

````python
if MENU_SIDE.lower() == "left":
    menu_position_css = "left: 24px;"
else:
    menu_position_css = "right: 24px;"


# For dark themes, use the card color as the menu background.
if THEME == "midnight":
    menu_background = CARD_BACKGROUND
else:
    menu_background = "rgba(255, 255, 255, 0.96)"


css = f'''
* {{
    box-sizing: border-box;
}}

html {{
    scroll-behavior: smooth;
}}

body {{
    margin: 0;
    font-family: {FONT_FAMILY};
    line-height: 1.65;
    color: {TEXT_COLOR};
    background: {BACKGROUND_COLOR};
}}

.container {{
    width: min({PAGE_WIDTH}, 84%);
    margin: 0 auto;
}}


/* Header */

header {{
    border-bottom: 1px solid {BORDER_COLOR};
    background: {BACKGROUND_COLOR};
    padding: 26px 0;
}}

header h1 {{
    margin: 0;
    font-size: 1.8rem;
}}

.subtitle {{
    margin: 3px 0 0;
    color: {MUTED_TEXT_COLOR};
}}


/* Fixed menu */

.side-nav {{
    position: fixed;
    top: 50%;
    {menu_position_css}
    transform: translateY(-50%);

    display: flex;
    flex-direction: column;
    gap: 10px;

    padding: 14px 12px;

    border: 1px solid {BORDER_COLOR};
    border-radius: 8px;

    background: {menu_background};

    z-index: 1000;
}}

.side-nav a {{
    color: {TEXT_COLOR};
    text-decoration: none;
    font-size: 0.9rem;
    padding: 5px 8px;
    border-radius: 4px;
}}

.side-nav a:hover {{
    background: {CARD_BACKGROUND};
}}


/* Sections */

main {{
    padding-bottom: 40px;
}}

section {{
    padding: 55px 0;
    border-bottom: 1px solid {BORDER_COLOR};
    scroll-margin-top: 30px;
}}

section h2 {{
    margin-top: 0;
    font-size: 1.7rem;
}}


/* Hero */

.hero {{
    padding-top: 70px;
    padding-bottom: 70px;
}}

.hero-text {{
    max-width: 720px;
}}

.hero h2 {{
    font-size: 2.5rem;
    line-height: 1.2;
    margin: 8px 0 20px;
}}

.eyebrow {{
    text-transform: uppercase;
    letter-spacing: 0.12em;
    font-size: 0.8rem;
    font-weight: bold;
    color: {MUTED_TEXT_COLOR};
}}

.button {{
    display: inline-block;
    margin-top: 10px;
    padding: 10px 18px;

    border: 1px solid {ACCENT_COLOR};

    color: {ACCENT_COLOR};
    text-decoration: none;
}}

.button:hover {{
    background: {ACCENT_COLOR};
    color: {BACKGROUND_COLOR};
}}


/* Cards */

.card-grid {{
    display: grid;
    grid-template-columns: repeat({CARD_COLUMNS}, 1fr);
    gap: 20px;
}}

.card {{
    padding: 22px;
    border: 1px solid {BORDER_COLOR};
    border-radius: 6px;
    background: {CARD_BACKGROUND};
}}

.card h3 {{
    margin-top: 0;
}}


/* Lists */

.clean-list {{
    list-style: none;
    padding: 0;
}}

.clean-list li {{
    padding: 15px 0;
    border-bottom: 1px solid {BORDER_COLOR};
}}


/* Links */

a {{
    color: {ACCENT_COLOR};
}}


/* Footer */

footer {{
    padding: 28px 0;
    color: {MUTED_TEXT_COLOR};
    font-size: 0.9rem;
}}


/* Small screens */

@media (max-width: 760px) {{

    .container {{
        width: 90%;
    }}

    .side-nav {{
        position: static;
        transform: none;

        flex-direction: row;
        flex-wrap: wrap;
        justify-content: center;

        border: none;
        border-bottom: 1px solid {BORDER_COLOR};
        border-radius: 0;

        padding: 12px;
    }}

    .hero h2 {{
        font-size: 2rem;
    }}

    .card-grid {{
        grid-template-columns: 1fr;
    }}
}}
'''

CSS_FILE.write_text(css, encoding="utf-8")

print("Created:")
print(CSS_FILE.resolve())
````

## 6. Preview the homepage

````python
from IPython.display import HTML, display

preview_html = html.replace(
    '<link rel="stylesheet" href="style.css">',
    f'<style>{css}</style>'
)

display(HTML(preview_html))
````

a. Lasso uses an L1 penalty, where lambda is multiplied by the sum of the absolute value of the coefficients and added to the loss function to penalize larger coefficients; with this method, the coefficients can reach zero. The elastic net method combines the ridge and lasso methods by multiplying the lasso penalty by a factor alpha and the ridge penalty by a factor 1 - alpha, adding them both to the loss function as a penalty for large coefficients. Since the elastic net method combines the lasso and ridge methods, it can set coefficients to zero.

b. When lambda is zero, the loss function is basically OLS because the penalty of either of the methods becomes 0, thus being taken out of the equation. When lambda increases, the penalty for the coefficients increases and the coefficients infinitely approach zero for the ridge method, and reach zero for the elastic net and lasso methods. Because the penalties for the coefficients all have the same weight controlled by the hyperparameter lambda, the predictors must be standardized before regularization. That is, if the predictors are not on the same scale as each other, then the penalty will be weighed differently for different predictors, making them uncomparable and more difficult to interpret. Thus, the predictors should be standardized to the same scale before being regularized.
